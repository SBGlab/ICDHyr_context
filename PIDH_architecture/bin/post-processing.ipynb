```python
import pandas as pd
from ete3 import Tree, TreeStyle, NodeStyle, TextFace, RectFace, CircleFace
from ete3 import NCBITaxa

# Load your file
df = filtered_df.drop_duplicates(subset=["taxid", "architecture_type"]) # adjust separator if needed

# Ensure taxid is int
df["taxid"] = df["taxid"].astype(int)

# Initialize taxonomy
ncbi = NCBITaxa()

# Get topology (this builds the phylogenetic tree)
taxids = df["taxid"].unique().tolist()
tree = ncbi.get_topology(taxids)

# Map taxid to architecture_type
arch_map = dict(zip(df["taxid"], df["architecture_type"]))

arch_colors = {
    "dimeric": "salmon",
    "monomeric": "mediumturquoise",
    "both": "MediumOrchid"
}

# Map habitat_class to taxid
habitat_map = dict(zip(df["taxid"], df["habitat_class"]))

habitat_colors = {
    "soil": "chocolate",
    "aquatic": "blue",
    "animal": "orangered",
    "plant": "green",
    "unknown": "gray"  
}

#Map taxid to species name (optional, for better labels)
name_map = dict(zip(df["taxid"], df["specie"]))

# Define branch colors
color_map = habitat_colors

def layout(node):
    if node.is_leaf():
        specie = name_map.get(int(node.name), node.name)
        arch = arch_map.get(int(node.name), "unknown")

        # TEXT LABEL
        label = TextFace(f"{specie}", fsize=8)
        node.add_face(label, column=0, position="aligned")

        # SYMBOL
        symbol_color = arch_colors.get(arch, "gray")

        symbol = CircleFace(radius=10, color=symbol_color, style="circle") 
        node.add_face(symbol, column=1, position="aligned")


# Traverse tree and color branches
for node in tree.traverse():
    nstyle = NodeStyle()

    if node.is_leaf():
        taxid = int(node.name)
        arch = habitat_map.get(taxid, "unknown")
        color = color_map.get(arch, "black")

        # Color the BRANCHES
        nstyle["hz_line_color"] = color
        nstyle["vt_line_color"] = color

        # Optional: node dot
        nstyle["fgcolor"] = color
        nstyle["size"] = 6

    else:
        # Internal branches (optional: keep black or compute consensus)
        nstyle["hz_line_color"] = "black"
        nstyle["vt_line_color"] = "black"
        nstyle["size"] = 0

    node.set_style(nstyle)

# Tree visualization
ts = TreeStyle()
# Circular mode
ts.mode = "c"

# Optional: improve readability
ts.show_leaf_name = False
ts.scale = 100  # controls spacing
ts.layout_fn = layout

# Title (fixed version)
ts.title.add_face(TextFace("IDH Architecture Phylogeny", fsize=14), column=0)

tree.show(tree_style=ts)

# Save outputs
tree.write(outfile="idh_tree.nw")
tree.render("idh_tree.png", tree_style=ts, w=1200)
```

```python
from ete3 import NCBITaxa

df = filtered_df

ncbi = NCBITaxa()
taxids = df["taxid"].unique().tolist()

lineages = ncbi.get_lineage_translator(taxids)
ranks = ncbi.get_rank(taxids)

def get_family(taxid):
    lineage = ncbi.get_lineage(taxid)
    names = ncbi.get_taxid_translator(lineage)
    rank_dict = ncbi.get_rank(lineage)

    for tid in lineage:
        if rank_dict.get(tid) == "family":
            return names.get(tid, None)
    return None

df["family"] = df["taxid"].apply(get_family)

family_df = df.groupby("family")

arch_comp = (
    df.groupby(["family", "architecture_type"])
    .size()
    .unstack(fill_value=0)
)

arch_comp = arch_comp.div(arch_comp.sum(axis=1), axis=0)

hab_comp = (
    df.groupby(["family", "habitat_class"])
    .size()
    .unstack(fill_value=0)
)

# IMPORTANT: avoid divide-by-zero collapse
row_sums = hab_comp.sum(axis=1)
hab_comp = hab_comp.div(row_sums.replace(0, 1), axis=0)
```

### Class-Level Taxonomy Summary

This section collapses the protein-level data to the bacterial class level, providing a high-level overview of IDH architecture distribution across major prokaryotic taxa. Each bacterial class is represented by one representative taxon for phylogenetic tree construction.

```python
# === CLASS-LEVEL EXTRACTION ===
# Extract bacterial class from NCBI taxonomy for each filtered taxon

from ete3 import NCBITaxa
import numpy as np

df_class = filtered_df.copy()
ncbi = NCBITaxa()
taxids_list = df_class["taxid"].unique().tolist()

def get_class(taxid):
    """Extract class rank from NCBI taxonomy lineage"""
    try:
        lineage = ncbi.get_lineage(int(taxid))
        names = ncbi.get_taxid_translator(lineage)
        rank_dict = ncbi.get_rank(lineage)
        
        for tid in lineage:
            if rank_dict.get(tid) == "class":
                return names.get(tid, None)
    except:
        pass
    return None

# Apply class extraction
df_class["class"] = df_class["taxid"].apply(get_class)

print("=" * 70)
print("CLASS-LEVEL TAXONOMY ANALYSIS")
print("=" * 70)

class_counts = df_class["class"].value_counts()
print(f"\nTotal unique bacterial classes identified: {len(class_counts)}")
print(f"\nClass distribution (number of species in each class):")
print(class_counts)

# Architecture distribution by class
arch_comp_by_class = df_class.groupby(["class", "architecture_type"]).size().unstack(fill_value=0)
print(f"\nArchitecture composition by class:")
print(arch_comp_by_class)
```

```python
# === CLASS-LEVEL VISUALIZATION ===

# Bar chart of class diversity
fig = px.bar(
    x=class_counts.index,
    y=class_counts.values,
    labels={"x": "Bacterial Class", "y": "Number of Species"},
    title="Species Distribution Across Bacterial Classes",
    color=class_counts.values,
    color_continuous_scale="Viridis"
)
fig.update_xaxes(tickangle=-45)
fig.update_layout(height=500, showlegend=False)
fig.show()

# Architecture distribution heatmap by class
fig = px.imshow(
    arch_comp_by_class,
    labels=dict(x="Architecture Type", y="Bacterial Class", color="Count"),
    title="IDH Architecture Distribution by Bacterial Class",
    color_continuous_scale="YlOrRd",
    aspect="auto"
)
fig.update_layout(height=600)
fig.show()

# Stacked bar chart showing architecture proportions per class
arch_comp_normalized = arch_comp_by_class.div(arch_comp_by_class.sum(axis=1), axis=0)
fig = px.bar(
    arch_comp_normalized,
    barmode="stack",
    labels={"index": "Bacterial Class", "value": "Proportion", "variable": "Architecture"},
    title="Proportion of IDH Architecture Types by Bacterial Class",
    color_discrete_map={"monomeric": "mediumturquoise", "dimeric": "salmon", "both": "MediumOrchid", "other": "lightgray"}
)
fig.update_xaxes(tickangle=-45)
fig.update_layout(height=500)
fig.show()
```

```python
# Get composition tables
arch_comp_class = (
    df.groupby(["class", "architecture_type"])
    .size()
    .unstack(fill_value=0)
)
arch_comp_class = arch_comp_class.div(arch_comp_class.sum(axis=1), axis=0)

hab_comp_class = (
    df.groupby(["class", "habitat_class"])
    .size()
    .unstack(fill_value=0)
)
row_sums = hab_comp_class.sum(axis=1)
hab_comp_class = hab_comp_class.div(row_sums.replace(0, 1), axis=0)

# Get representative taxids for each class
rep_class = df.groupby("class")["taxid"].first().to_dict()
class_size = df.groupby("class").size().rename("n_taxa")

# Validate taxids
valid_taxids_class = []
for taxid in rep_class.values():
    try:
        ncbi.get_lineage(int(taxid))
        valid_taxids_class.append(int(taxid))
    except:
        continue

print(f"Total classes: {len(class_size)}")
print(f"Valid taxids: {len(valid_taxids_class)}")
print(class_size)
```

```python
# Build class-level tree
tree_class = ncbi.get_topology(valid_taxids_class)

inv_class = {taxid: cls for cls, taxid in rep_class.items() if taxid in valid_taxids_class}

for leaf in tree_class.iter_leaves():
    taxid = int(leaf.name)
    if taxid in inv_class:
        leaf.name = inv_class[taxid]
    else:
        leaf.name = "unmapped"

for node in tree_class.traverse():
    node.dist = 1

tree_class.write(format=9, outfile="class_tree_new_dimeric_threshold.nwk")

# Prepare data
arch_comp_class = arch_comp_class.fillna(0)
hab_comp_class = hab_comp_class.fillna(0)

arch_comp_class = arch_comp_class.reindex(sorted(arch_comp_class.columns), axis=1)
hab_comp_class = hab_comp_class.reindex(sorted(hab_comp_class.columns), axis=1)

arch_colors = {
    "monomeric": "#e41a1c",
    "dimeric": "#377eb8",
    "both": "#4daf4a",
    "unknown": "#999999"
}

hab_colors = {
    "soil": "#8c510a",
    "aquatic": "#1b9e77",
    "freshwater": "#a6cee3",
    "marine": "#1b249e",
    "animal": "#d73027",
    "extreme": "#6a3d9a",
    "mixed": "#bdbdbd",
    "unknown": "#969696"
}

hab_fields = list(hab_comp_class.columns)
hab_colors_list = [hab_colors.get(x, "#999999") for x in hab_fields]

arch_fields = list(arch_comp_class.columns)
arch_colors_list = [arch_colors.get(x, "#999999") for x in arch_fields]

# === iTOL DATASETS FOR CLASS ===

# Architecture dataset
with open("itol_architecture_class_new_dimeric_threshold.txt", "w") as f:
    f.write("DATASET_MULTIBAR\n")
    f.write("SEPARATOR TAB\n")
    f.write("DATASET_LABEL\tArchitecture composition (Class level)\n")
    f.write("COLOR\t#ff0000\n\n")
    f.write("FIELD_LABELS\t" + "\t".join(arch_fields) + "\n")
    f.write("FIELD_COLORS\t" + "\t".join(arch_colors_list) + "\n\n")
    f.write("WIDTH\t50\n")
    f.write("HEIGHT\t10\n")
    f.write("MARGIN\t1\n\n")
    f.write("DATA\n")
    for cls, row in arch_comp_class.iterrows():
        f.write(f"{cls}\t" + "\t".join(map(str, row[arch_fields].values)) + "\n")

# Habitat dataset
with open("itol_habitat_class_new_dimeric_threshold.txt", "w") as f:
    f.write("DATASET_MULTIBAR\n")
    f.write("SEPARATOR TAB\n")
    f.write("DATASET_LABEL\tHabitat composition (Class level)\n")
    f.write("COLOR\t#00aa00\n\n")
    f.write("FIELD_LABELS\t" + "\t".join(hab_fields) + "\n")
    f.write("FIELD_COLORS\t" + "\t".join(hab_colors_list) + "\n\n")
    f.write("WIDTH\t50\n")
    f.write("HEIGHT\t10\n")
    f.write("MARGIN\t5\n\n")
    f.write("DATA\n")
    for cls, row in hab_comp_class.iterrows():
        values = "\t".join(map(str, row[hab_fields].values))
        f.write(f"{cls}\t{values}\n")

# Abundance dataset
# Min-Max with floor: scale to 1-100 range (never 0)
#scaled_class = (class_size - class_size.min()) / (class_size.max() - class_size.min()) * 99 + 1
scaled_class = class_size

with open("itol_abundance_class_new_dimeric_threshold.txt", "w") as f:
    f.write("DATASET_SIMPLEBAR\n")
    f.write("SEPARATOR TAB\n")
    f.write("DATASET_LABEL\tClass size\n")
    f.write("COLOR\t#000000\n\n")
    f.write("WIDTH\t50\n")
    f.write("HEIGHT\t50\n")
    f.write("MARGIN\t1\n\n")
    f.write("DATA\n")
    for cls, val in scaled_class.items():
        f.write(f"{cls}\t{val}\n")

# === Kingdom dataset for class-level tree ===
# Create composition table grouped by class and kingdom
kingdom_comp_class = (
    df.groupby(["class", "kingdom"])
    .size()
    .unstack(fill_value=0)
)
# Normalize by class totals (proportion per class)
kingdom_comp_class = kingdom_comp_class.div(kingdom_comp_class.sum(axis=1), axis=0)

# Get kingdom field names
kingdom_fields = list(kingdom_comp_class.columns)

# Generate dynamic colors for kingdom categories from extended palette
base_kingdom_colorscale = px.colors.qualitative.Plotly + px.colors.qualitative.Set1 + px.colors.qualitative.Set2
kingdom_colors_list = [base_kingdom_colorscale[i % len(base_kingdom_colorscale)] for i in range(len(kingdom_fields))]

# Write iTOL DATASET_MULTIBAR file for kingdom composition
with open("itol_kingdom_class_new_dimeric_threshold.txt", "w") as f:
    f.write("DATASET_MULTIBAR\n")
    f.write("SEPARATOR TAB\n")
    f.write("DATASET_LABEL\tKingdom composition (Class level)\n")
    f.write("COLOR\t#0000ff\n\n")
    f.write("FIELD_LABELS\t" + "\t".join(kingdom_fields) + "\n")
    f.write("FIELD_COLORS\t" + "\t".join(kingdom_colors_list) + "\n\n")
    f.write("WIDTH\t50\n")
    f.write("HEIGHT\t10\n")
    f.write("MARGIN\t1\n\n")
    f.write("DATA\n")
    for cls, row in kingdom_comp_class.iterrows():
        f.write(f"{cls}\t" + "\t".join(map(str, row[kingdom_fields].values)) + "\n")

print("Kingdom dataset created: itol_kingdom_class_new_dimeric_threshold.txt")

print("\nClass-level files generated:")
print("- class_tree_new_dimeric_threshold.nwk")
print("- itol_architecture_class_new_dimeric_threshold.txt")
print("- itol_habitat_class_new_dimeric_threshold.txt")
print("- itol_abundance_class_new_dimeric_threshold.txt")
print("- itol_kingdom_class_new_dimeric_threshold.txt")
```
